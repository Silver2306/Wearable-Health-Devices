```python
#a statement to see if the issue is resolved of me and masboi
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Model performance
models = ["Logistic Regression", "SVM", "Random Forest", "XGBoost"]

accuracy = [0.7136, 0.7059, 0.6938, 0.7080]
precision = [0.71, 0.71, 0.69, 0.71]
recall = [0.71, 0.70, 0.69, 0.71]
f1 = [0.71, 0.71, 0.69, 0.71]

results = pd.DataFrame({
    "Model": models,
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1
})

results
```

```python
def compare_two(model1, model2):

    selected = results[results["Model"].isin([model1, model2])]

    metrics = ["Accuracy", "Precision", "Recall", "F1 Score"]

    x = np.arange(len(metrics))
    width = 0.35

    plt.figure(figsize=(9, 5))

    plt.bar(x - width/2,
            selected.iloc[0][metrics],
            width,
            label=model1)

    plt.bar(x + width/2,
            selected.iloc[1][metrics],
            width,
            label=model2)

    plt.xticks(x, metrics)
    plt.ylim(0, 1)
    plt.ylabel("Score")
    plt.title(f"{model1}  VS  {model2}")
    plt.legend()

    plt.show()
```

```python
#LR vs SVM
compare_two("Logistic Regression", "SVM")
```

```python
#LR vs RF
compare_two("Logistic Regression", "Random Forest")
```

```python
#LR VS XG
compare_two("Logistic Regression", "XGBoost")
```

```python
#SVM VS RF
compare_two("SVM", "Random Forest")
```

```python
#SVM vS XG
compare_two("SVM", "XGBoost")
```

```python
#RF VS XG
compare_two("Random Forest", "XGBoost")
```

```python
metrics = ["Accuracy", "Precision", "Recall", "F1 Score"]

x = np.arange(len(metrics))
width = 0.2

plt.figure(figsize=(12, 6))

for i, model in enumerate(models):
    values = results[results["Model"] == model][metrics].values[0]

    plt.bar(
        x + (i - 1.5) * width,
        values,
        width,
        label=model
    )

plt.xticks(x, metrics)
plt.ylim(0, 1)
plt.ylabel("Score")
plt.title("Comparison of All Four Classification Models")
plt.legend()

plt.show()
```

```python
#Class wise f1 comparison
class_f1 = pd.DataFrame({
    "Model": models,
    "High": [0.78, 0.77, 0.76, 0.77],
    "Low": [0.78, 0.78, 0.77, 0.78],
    "Moderate": [0.58, 0.57, 0.54, 0.57]
})

x = np.arange(3)
width = 0.2

plt.figure(figsize=(10, 6))

for i, model in enumerate(models):
    values = class_f1[class_f1["Model"] == model][
        ["High", "Low", "Moderate"]
    ].values[0]

    plt.bar(
        x + (i - 1.5) * width,
        values,
        width,
        label=model
    )

plt.xticks(x, ["High", "Low", "Moderate"])
plt.ylim(0, 1)
plt.ylabel("F1 Score")
plt.xlabel("Stress Class")
plt.title("Class-wise F1 Score Comparison")
plt.legend()

plt.show()
```
